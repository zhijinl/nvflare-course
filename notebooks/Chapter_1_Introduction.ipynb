# What is Federated Learning

Data is the cornerstone of developing generalizable and robust AI models. However, collecting a representative pool of real-world data for AI training and validation continues to pose significant challenges. These obstacles include regulatory constraints, limited data availability, and privacy concerns.

Let's examine an example from the healthcare domain to illustrate these difficulties. A [recent study](https://pmc.ncbi.nlm.nih.gov/articles/PMC6389801/) revealed that among 516 published medical AI studies in 2018, only 6% performed external validations. Even more concerning, an even smaller fraction of these studies included multi-institutional and prospective validation data. This phenomenon can be largely attributed to the regulatory, legal, and technical hurdles in accessing high-quality, multi-institutional medical data for traditional AI development approaches.

Building real-world AI applications requires a fundamental shift in our approach to data collection, sharing, and utilization.

![Real World AI](../images/real-world-ai.png)

[Federated Learning (FL)](https://en.wikipedia.org/wiki/Federated_learning) is a distributed, collaborative machine learning paradigm which potentially solves these data access challenges. In FL, instead of centralizing local datasets to train a model, copies of a global model are sent to different sites and the global model is trained collaboratively in an iterative way: the training steps occur at each site locally, with its proprietary datasets, and only the insights, not the data itself, are aggregated to update the global model before re-distribution again to the local sites.

Compared with centralized learning, FL enables the creation of commonly robust models while holding multiple benefits:
- **Enhanced data privacy and security**: FL facilitates data privacy and data locality by ensuring that the data remains at each site. Additionally, privacy preserving techniques such as homomorphic encryption and differential privacy filters can be further leveraged to protect the transferred data.
- **Improved model performance and diversity**: by training with a variety of external data sources across different clients, a robust and generalizable global model can be developed to better represent a heterogeneous data distribution in a real-world scenario.
- **Scalability and efficiency**: by performing training locally at each site, federated learning can scale effectively across global networks. Since only model weights need to be transferred instead of entire datasets, it also makes efficient use of network resources.

Typically, an FL project with a client-server architecture would include the following concepts:
- **FL server**: a central entity that manages job lifecycle, orchestrates the workflow, assigns training / computation tasks to clients and performs aggregation of computation results received from clients.
- **FL client**: entity representing a local site. A client executes tasks assigned by the FL server, performs local computation / training with local dataset, and submits result back to FL server.
- **FL workflow**: an FL workflow is a distributed workflow with specific optimization algorithms and communication strategies, which determine how local training / computations are performed and how results aggregation is performed on the server side.

There are various types of FL. In the centralized version of FL, the FL server has the role of the aggregrator node. However, in a decentralized version such as [Swarm Learning](https://www.nature.com/articles/s41586-021-03583-3), FL clients can serve as the aggregator nodes instead. An FL workflow can be horizontal, in the sense that clients hold different data samples with the same features; while in vertical FL, clients hold different features over an overlapping set of data samples.

![FL centralized](../images/fl-diagram.png)

In a typical centralized FL, the server orchestrates the collaboration of multiple clients by first sending an initial model to the FL clients. The clients perform a few training steps on their local datasets, then send the model updates back to the FL server for aggregation to update a global model. This process forms a single round of federated learning and after a number of rounds, a robust global model can be developed. A classic example is the [Federated Averaging](https://arxiv.org/pdf/1602.05629) (FedAvg) algorithm, where the server aggregates the received model updates from clients by performing a weighted average, to obtain the global model. While being a straight-forward algorithm, FedAvg does not account for data heterogeneity amoung different clients. Over recent years, many other federated algorithms have been proposed to better deal with data heterogeneity. Popular ones include [FedOpt](https://arxiv.org/abs/2003.00295), [FedProx](https://arxiv.org/abs/1812.06127), [Scaffold](https://arxiv.org/abs/1910.06378) etc.

![Federated Averaging](../images/fedavg.png)

Today, FL has applications over various industry sectors. For instance, an important application of FL is in the healthcare sector, where data privacy regulations and patient record confidentiality make training models challenging. FL can help break down these healthcare data silos to allow hospitals and medical institutions to collaborate and pool their medical knowledge without the need to share their data. Some common use cases involve cancer classification and detection, drug discovery with federated protein large language models, federated analytics of medical devices etc. Furthermore, FL may beneficial to many other areas and industries such as financial fraud detection, autonomous vehicles, HPC, mobile applications, etc., where the ability to use distributed data silos while maintaining data privacy is essential for the development of better models.


# What is NIVDIA FLARE

Implementing FL is a challenging task. First, carefully designed software tools are needed for scientists and developers to quickly and easily prototype an FL project and test its feasibility. Additionally, tools to properly perform project monitoring, experiment tracking and debugging are also needed to effectively design and build FL projects. Next, there are still open research questions regarding the performance of FL in comparison with centralized learning paradigm, especially for heterogeneous data distributions. Implementation of an FL workflow needs to be flexible enough to account for new distributed optimization algorithms that are being proposed. Then, to build real-world AI leveraging FL, a proper implementation of security and privacy measures is needed. These measures include for instance user authentication, authorization, data encryption and privacy filters etc. Finally, in real-world FL deployment, a production-ready and scalable runtime  environment is also needed to handle aspects such as service high availability (back-up servers), mixture of compute resources and concurrent job executions etc.

NVIDIA Federated Learning Application Runtime Environment (NVIDIA FLARE) was created to tackle the various challenges in implementing FL. NVIDIA FLARE is a **domain- & framework-agnostic, open-source (apache 2 license), extensible Python SDK** that allows researchers and data scientists to easily adapt existing machine / deep learning workflow to a federated paradigm. NVIDIA FLARE SDK is also built for robust and production scale real-world FL deployments. With NVIDIA FLARE, developers can build a secure, privacy preserving offering for a distributed multi-party collaboration. 

Let us quick review the core components of NVIDIA FLARE.

![NVIDIA FLARE](../images/FLARE-platform.jpg)

NVIDIA FLARE includes carefully designed APIs for data scientists and researchers to easily adapt existing computation / training code to a federated paradigm. FLARE also provides runtime environment enabling data scientists and researchers to easily carry out FL experiments in a simulated environment, with extensible tools for project monitoring.

NVIDIA FLARE is built to be componentized, and highly customizable. It includes built-in implementations of common federated training workflows, such as scatter-and-gather, cyclic, etc., federated evaluation workflows such as global model evaluation, cross site model validation, etc., and learning algorithms such as FedAvg, FedOpt, FedProx etc. Researchers can easily extend basic components provided in FLARE to implement their own workflows and algorithms.

NVIDIA FLARE also offers features to bring FL from research to real-world deployment. It has exetensible tools for secure provisioning and orchestration. Its production-level runtime allows for large scale, concurrent multi-task execution. FLARE provides an extensible set of security features, such as privacy preserving technologies (homomorphic encryption, differential privacy, ...) and support for confidential computing. Also, NVIDIA FLARE offers system capabilities to set up FL with high availability infrastructure.

With this set of core components, NVIDIA FLARE aims to be a versatile and extensible SDK for both federated algorithm researchers and real-world federated application developers.
