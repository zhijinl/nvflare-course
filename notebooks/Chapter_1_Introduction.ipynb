# What is Federated Learning

Building real-world AI applications is challenging: it requires external, multi-institutional and prospective data. 

A [recent study](https://pmc.ncbi.nlm.nih.gov/articles/PMC6389801/) shows that, amoung 516 pubilished medical AI studies in 2018, only 6% of them performed external validations. Even fewer of these studies included multi-institutional and prospective validation data. Today, getting access to high quality multi-institutional data in traditional machine learning paradigm still poses many challenges in terms of data sharing regulation, legal constraints, privacy concerns, technical considerations etc. 

<img src="../images/real-world-ai.png" alt="Real World AI" width=35% height=35% />

Federated Learning (FL) is a distributed machine learning paradigm which potentially solves these data access challenges. In FL, learning occurs at each site locally, with its own local datasets. Instead of centralizing local datasets, a global model is obtained via aggregating locally trained models. Compared with centralized learning, FL enables the creation of commonly robust models while holding multiple benefits:
- **Enhanced Data Privacy and Security**: FL facilitates data privacy and data locality by ensuring that the data remains at each site. Additionally, privacy preserving techniques such as homomorphic encryption and differential privacy filters can be further leveraged to protect the transferred data.
- **Improved Model Performance and Diversity**: by training with a variety of external data sources across different clients, a robust and generalizable global model can be developed to better represent heterogeneous data distribution in real-world.
- **Scalability and Efficiency**: with the ability to perform training locally at the edge, federated learning can be highly scalable across the globe. Additionally, only needing to transfer the model weights rather than entire datasets enables efficient use of resources.

Typically, an FL project would include the following concepts:
- **FL server**: a central entity that manages job lifecycle, orchestrates the workflow, assigns training / computation tasks to clients and performs aggregation of computation results received from clients.
- **FL client**: entity representing a local site. A client executes tasks assigned by the FL server, performs local computation / training with local dataset, and submits result back to FL server.
- **FL workflow**: an FL workflow is a distributed workflow with specific optimization algorithms and communication strategies, which determine how local training / computations are performed and how results aggregation is performed on the server side.

There are various types of FL. In the centralized version of FL, the FL server has the role of the aggregrator node. However, in a decentralized version such as [swarm learning](https://www.nature.com/articles/s41586-021-03583-3), FL clients can serve as the aggregator nodes instead. An FL workflow can be horizontal, in the sense that clients hold different data samples with the same features; while in vertical FL, clients hold different features over an overlapping set of data samples.

<img src="../images/fl-diagram.png" alt="FL centralized" width=50% height=50% />

In a typical centralized FL, the server orchestrates the collaboration of multiple clients by first sending an initial model to the FL clients. The clients perform training on their local datasets, then send the model updates back to the FL server for aggregation to update a global model. This process forms a single round of federated learning and after a number of rounds, a robust global model can be developed. A classic example is the Federated Averaging (FedAvg) algorithm, where the server aggregates the received model updates from clients by performing a weighted average, to obtain the global model. While being a straight-forward algorithm, FedAvg does not account for data heterogeneity amoung different clients. Over recent years, many other federated algorithms have been proposed to better deal with data heterogeneity. Popular ones include FedOpt, FedProx, Scaffold etc.

Today, FL has applications over various industry sectors. For instance, an important application of federated learning is in the healthcare sector, where data privacy regulations and patient record confidentiality make training models challenging. Federated learning can help break down these healthcare data silos to allow hospitals and medical institutions to collaborate and pool their medical knowledge without the need to share their data. Some common use cases involve cancer classification and detection, drug discovery with federated protein large language models, federated analytics of medical devices etc. Furthermore, there are many other areas and industries such as financial fraud detection, autonomous vehicles, HPC, mobile applications, etc., where the ability to use distributed data silos while maintaining data privacy is essential for the development of better models.


# What is NIVDIA FLARE

Implementing FL is a challenging task. First, carefully designed software tools are needed for scientists and developers to quickly and easily prototype an FL project and test its feasibility. Additionally, tools to properly perform project monitoring, experiment tracking and debugging are also needed to effectively design and build FL projects. Next, there are still open research questions regarding the performance of FL in comparison with centralized learning paradigm, especially for heterogeneous data distributions. Implemetation of an FL workflow needs to be flexible enough to account for new distributed optimization algorithms that are constantly being proposed. Then, to build real-world AI leveraging FL, a proper implementation of security and privacy measures is needed. These measures include for instance user authentication, authorization, data encryption and privacy filters etc. Finally, in real-world FL deployment, a production-ready and scalable runtime  environment is also needed to handle aspects such as service high availability (back-up servers), mixture of compute resources and concurrent job executions etc.

NVIDIA Federated Learning Application Runtime Environment (NVIDIA FLARE) was created to tackle the various challenges in implementing FL. NVIDIA FLARE is a **domain- & framework-agnostic, open-source, extensible Python SDK** that allows researchers and data scientists to easily adapt existing machine / deep learning workflow to a federated paradigm. NVIDIA FLARE SDK is also built for robust and production scale real-world FL deployments. With NVIDIA FLARE, developers can build a secure, privacy preserving offering for a distributed multi-party collaboration. 

Let us quick review the core components of NVIDIA FLARE.

<img src="../images/FLARE-platform.jpg" alt="FLARE Platform" width=50% height=50% />

NVIDIA FLARE includes carefully designed APIs for data scientists and researchers to easily adapt existing computation / training code to a federated paradigm. FLARE also provides runtime environment enabling data scientists and researchers to easily carry out FL experiments in a simulated environment, with extensible tools for project monitoring.

NVIDIA FLARE is built to be componentized, and highly customizable. It includes built-in implementations of common federated training workflows, such as scatter-and-gather, cyclic, etc., federated evaluation workflows such as global model evaluation, cross site model validation, etc., and learning algorithms such as FedAvg, FedOpt, FedProx etc. Researchers can easily extend basic components provided in FLARE to implement their own workflows and algorithms.

NVIDIA FLARE also offers features to bring FL from research to real-world deployment. It has exetensible tools for secure provisioning and orchestration. Its production-level runtime allows for large scale, concurrent multi-task execution. FLARE provides an extensible set of security features, such as privacy preserving technologies (homomorphic encryption, differential privacy, ...) and support for Confidential Computing. Also, FLARE offers system capabilities to set up federated learning with high availability infrastructure.

With this set of core components, NVIDIA FLARE aims to be a versatile and extensible SDK for both federated algorithm researchers and real-world federated application developers.
