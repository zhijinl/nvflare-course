## Intro to Provisioning with NVIDIA FLARE

In addition to the FL Simulator and POC modes, NVFlare ships with a provisioning system that can be used to generate scripts and configuration files for secure deployment of the FL server and clients.  These configurations also include certificates for establishing identity and secure communication between participants, and are packaged in unique subdirectory for each client and server.  We call these packages "startup kits".

To get started with the provisioning system, we can look at the `nvflare provision` command.


````python
!nvflare provision -h
````

Provisioning is based on a PROJECT_FILE that defines the server and clients, and the set of NVFlare Builder modules that are used to generate the startup kits.  You can run `nvflare provision` without any arguments to generate a sample `project.yml` file.


### Provisioning for Distributed Deployment

For this demo, we've already generated a `project.yml` config and have added an additional Builder module to generate a Docker compose environment for testing a deployment locally.  This configuration is in [`files/project.yml`](files/project.yml) (click to open in an editor).  The DockerBuilder module was added following the WorkspaceBuilder config with:
```yaml
  - path: nvflare.lighter.impl.docker.DockerBuilder
    args:
      base_image: flare-lab-client:latest
```

We won't do it here, but we could use the `nvflare_compose` configuration generated during provisioning to run a set of containers to test the server and client configs.  All that's required is building the docker image specified above, for example, by running `docker build -f /flare/notebooks/files/Dockerfile -t flare-lab-client .`  You can then edit the compose.yaml and .env files in the `nvflare_compose` configuration and launch a the set of server and client containers using `docker compose up`.

To generate the provisioning workspace, just run `nvflare provision -w example_workspace -p files/project.yml`:

### The Provisioning Workspace

When we run provisioning for a secure, distributed deployment, you'll notice a structure similar to that in POC mode, but with additional files for identity and authorization policies.  These includes a set of self signed certificates that allow clients to authenticate with the server, along with `local` directories for each participant that allow individual sites to customize their logging, privacy, and authorization policies.  For a distributed deployment, the project lead would typically distribute these client and server directories to the institutions participating in the project as encrypted zip files.
