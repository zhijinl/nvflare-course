# Federated Learning

Federated Learning (FL) is a distributed learning paradigm where training occurs at multiple local sites, each with their own local datasets. The global model is obtained via aggregating locally trained models. Compared with centralized training paradigm, FL enables the creation of common robust models without sharing sensitive local data, helping solve issues of data privacy and security. 

Typically, an FL project would include the following concepts:
- **FL server**: a central node that manages job lifecycle, orchestrates the workflow, assigns training / computation tasks to clients and performs aggregation of computation results received from clients.
- **FL client**: node representing a local site. A client executes tasks assigned by the FL server, performs local computation / training with local dataset, and submits result back to FL server.
- **FL workflow**: an FL workflow is a distributed workflow with specific optimization algorithms and communication strategies, which determine how local training / computations are performed and how results aggregation is performed on the server side.

There are various types of FL. In the centralized version of FL, the FL server has the role of the aggregrator node. However, in a decentralized version such as swarm learning, FL clients can serve as the aggregator node instead. An FL workflow can be horizontal, in the sense that clients hold different data samples with the same features; while in vertical FL, clients hold different features over an overlapping set of data samples.

In a typical centralized FL, the server orchestrates the collaboration of multiple clients by first sending an initial model to the FL clients. The clients perform training on their local datasets, then send the model updates back to the FL server for aggregation to form a global model. This process forms a single round of federated learning and after a number of rounds, a robust global model can be developed. A classic example is the Federated Averaging (FedAvg) algorithm, where during local training, each client would come up with its own model weights or weight updates, and send them to the server. The server then aggregates the received model weights or weight updates from all clients by simply performing a weighted average, to obtain the global model. While being a straight-forward algorithm, FedAvg does not account for data heterogeneity amoung different clients. Over recent years, many other federated algorithms have been proposed to better deal with data heterogeneity. Popular ones include FedOpt, FedProx, Scaffold etc.

FL holds multiple benefits over traditional learning paradigms:
- **Enhanced Data Privacy and Security**: FL facilitates data privacy and data locality by ensuring that the data remains at each site. Additionally, privacy preserving techniques such as homomorphic encryption and differential privacy filters can also be leveraged to further protect the transferred data.
- **Improved Accuracy and Diversity**: by training with a variety of data sources across different clients, a robust and generalizable global model can be developed to better represent heterogeneous datasets.
- **Scalability and Network Efficiency**: with the ability to perform training at the edge, federated learning can be highly scalable across the globe. Additionally, only needing to transfer the model weights rather than entire datasets enables efficient use of resources.

Today, FL has applications over various industry sectors. For instance, an important application of federated learning is in the healthcare sector, where data privacy regulations and patient record confidentiality make training models challenging. Federated learning can help break down these healthcare data silos to allow hospitals and medical institutions to collaborate and pool their medical knowledge without the need to share their data. Some common use cases involve cancer classification and detection, drug discovery with federated protein LLMs, federated analytics on medical devices etc. Furthermore, there are many other areas and industries such as financial fraud detection, autonomous vehicles, HPC, mobile applications, etc., where the ability to use distributed data silos while maintaining data privacy is essential for the development of better models.

# NVDIAI FLARE

NVIDIA FLARE (NVIDIA Federated Learning Application Runtime Environment) is a domain- & framework-agnostic, open-source, extensible python SDK that allows researchers and data scientists to easily adapt existing machine / deep learning workflow to a federated paradigm. With NVIDIA FLARE, platform developers can build a secure, privacy preserving offering for a distributed multi-party collaboration. NVIDIA FLARE SDK is built for robust and production scale real-world federated learning deployments.

<img src="../images/nvflare.png" alt="NVFLARE" width=50% height=50% />

Core components of NVIDIA FLARE include:
- Carefully designed APIs for data scientists and researchers to easily adapt existing computation / training code to a federated paradigm.
- Highly customizable components, with built-in implementations of:
  - Federated training workflows (scatter-and-gather, cyclic, etc.)
  - Federated evaluation workflows (global model evaluation, cross site model validation);
  - Learning algorithms (FedAvg, FedOpt, FedProx, etc.)
  - Privacy preserving algorithms (homomorphic encryption, differential privacy)
- A rich set of programmable APIs allowing researchers to create new federated workflows, learning & privacy preserving algorithms.
- A runtime environment enabling data scientists and researchers to easily carry out FL experiments in a simulated environment, as well as real-world scenario. NVIDIA FLARE supports multiple task execution, maximizing data scientist’s productivity.
- System capabilities to set up federated learning with high availability infrastructure.
- Extensible management tools for:
  - Secured provisioning
  - Orchestration
  - Monitoring of federated learning experiments
